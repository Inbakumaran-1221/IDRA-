```python
import pandas as pd

# Load the dataset
file_name = "Day8_Ecommerce_Sales_Dataset.csv"
df = pd.read_csv(file_name)

print("E-Commerce Sales Dataset")
display(df.head())


# 1. Basic Information

print("Dataset Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)


# 2. Data Selection

print("\nSelected Columns:")
display(df[["Order_ID", "Customer", "Product", "Total_Sales"]].head())

print("\nDetails of First Order:")
display(df.iloc[0])


# 3. Filtering Data

print("\nOrders with Sales Greater Than 5000:")
high_sales = df[df["Total_Sales"] > 5000]
display(high_sales)

print("\nOrders with Quantity Greater Than 3:")
high_quantity = df[df["Quantity"] > 3]
display(high_quantity)

print("\nUPI Payment Orders:")
upi_orders = df[df["Payment_Method"] == "UPI"]
display(upi_orders)

print("\nElectronics Category Orders:")
electronics_orders = df[df["Category"] == "Electronics"]
display(electronics_orders)


# 4. Sorting Data

print("\nTop 10 Orders by Sales:")
top_orders = df.sort_values("Total_Sales", ascending=False).head(10)
display(top_orders[["Order_ID", "Product", "Category", "Total_Sales"]])

print("\nOrders with Lowest Sales:")
lowest_orders = df.sort_values("Total_Sales").head(10)
display(lowest_orders[["Order_ID", "Product", "Category", "Total_Sales"]])

print("\nTop Products by Quantity Sold:")
top_quantity = df.sort_values("Quantity", ascending=False).head(10)
display(top_quantity[["Product", "Quantity", "Total_Sales"]])


# 5. Overall Sales Statistics

total_sales = df["Total_Sales"].sum()
average_sales = df["Total_Sales"].mean()
highest_sales = df["Total_Sales"].max()
lowest_sales = df["Total_Sales"].min()
total_quantity = df["Quantity"].sum()
number_of_orders = df["Order_ID"].nunique()

print("\n===== Overall Sales Statistics =====")
print(f"Total Sales: ₹{total_sales:.2f}")
print(f"Average Sales per Order: ₹{average_sales:.2f}")
print(f"Highest Sale: ₹{highest_sales:.2f}")
print(f"Lowest Sale: ₹{lowest_sales:.2f}")
print(f"Total Quantity Sold: {total_quantity}")
print(f"Number of Orders: {number_of_orders}")


# 6. Category Wise Analysis

category_sales = df.groupby("Category")["Total_Sales"].agg(
    ["sum", "mean", "count"]
).sort_values("sum", ascending=False)

category_sales.columns = ["Total Sales", "Average Sales", "Number of Orders"]

print("\n===== Category Wise Sales =====")
display(category_sales)

category_quantity = df.groupby("Category")["Quantity"].sum().sort_values(
    ascending=False
)

print("\n===== Category Wise Quantity Sold =====")
display(category_quantity)


# 7. City Wise Analysis

city_sales = df.groupby("City")["Total_Sales"].agg(
    ["sum", "mean", "count"]
).sort_values("sum", ascending=False)

city_sales.columns = ["Total Sales", "Average Sales", "Number of Orders"]

print("\n===== City Wise Sales =====")
display(city_sales)


# 8. Product Wise Analysis

product_sales = df.groupby("Product")["Total_Sales"].agg(
    ["sum", "mean", "count"]
).sort_values("sum", ascending=False)

product_sales.columns = ["Total Sales", "Average Sales", "Number of Orders"]

print("\n===== Product Wise Sales =====")
display(product_sales)


# Top 10 Products
print("\n===== Top 10 Products by Sales =====")
display(product_sales.head(10))


# 9. Payment Method Analysis

payment_sales = df.groupby("Payment_Method")["Total_Sales"].agg(
    ["sum", "mean", "count"]
).sort_values("sum", ascending=False)

payment_sales.columns = ["Total Sales", "Average Sales", "Number of Orders"]

print("\n===== Payment Method Analysis =====")
display(payment_sales)


# 10. Region Wise Analysis

region_sales = df.groupby("Region")["Total_Sales"].agg(
    ["sum", "mean", "count"]
).sort_values("sum", ascending=False)

region_sales.columns = ["Total Sales", "Average Sales", "Number of Orders"]

print("\n===== Region Wise Sales =====")
display(region_sales)


# 11. Find Specific Orders

print("\nOrders with Sales Between ₹3000 and ₹7000:")
selected_orders = df[
    (df["Total_Sales"] >= 3000) &
    (df["Total_Sales"] <= 7000)
]

display(
    selected_orders[
        ["Order_ID", "Customer", "Product", "Total_Sales"]
    ]
)


print("\nOrders with Rating Above 4.5:")
high_rated_orders = df[df["Rating"] > 4.5]

display(
    high_rated_orders[
        ["Order_ID", "Customer", "Product", "Rating", "Total_Sales"]
    ]
)


# 12. Highest Selling Order

highest_order = df.loc[df["Total_Sales"].idxmax()]

print("\n===== Highest Selling Order =====")
print(f"Order ID: {highest_order['Order_ID']}")
print(f"Customer: {highest_order['Customer']}")
print(f"Product: {highest_order['Product']}")
print(f"Category: {highest_order['Category']}")
print(f"City: {highest_order['City']}")
print(f"Sales: ₹{highest_order['Total_Sales']:.2f}")


# 13. Product with Highest Total Sales

best_product = df.groupby("Product")["Total_Sales"].sum().idxmax()
best_product_sales = df.groupby("Product")["Total_Sales"].sum().max()

print("\n===== Best Performing Product =====")
print(f"Product: {best_product}")
print(f"Total Sales: ₹{best_product_sales:.2f}")


# 14. Best Performing Category

best_category = df.groupby("Category")["Total_Sales"].sum().idxmax()
best_category_sales = df.groupby("Category")["Total_Sales"].sum().max()

print("\n===== Best Performing Category =====")
print(f"Category: {best_category}")
print(f"Total Sales: ₹{best_category_sales:.2f}")


# 15. Most Used Payment Method

popular_payment = df["Payment_Method"].value_counts().idxmax()
payment_count = df["Payment_Method"].value_counts().max()

print("\n===== Most Used Payment Method =====")
print(f"Payment Method: {popular_payment}")
print(f"Number of Orders: {payment_count}")


# 16. Basic Observations

city_total = df.groupby("City")["Total_Sales"].sum()
top_city = city_total.idxmax()

category_total = df.groupby("Category")["Total_Sales"].sum()
top_category = category_total.idxmax()

product_total = df.groupby("Product")["Total_Sales"].sum()
top_product = product_total.idxmax()

payment_count = df["Payment_Method"].value_counts()
top_payment = payment_count.idxmax()

print("\n========== KEY OBSERVATIONS ==========")

print(f"1. The dataset contains {len(df)} orders and {len(df.columns)} columns.")

print(
    f"2. Total sales generated from all orders is "
    f"₹{total_sales:.2f}."
)

print(
    f"3. The highest performing category based on total sales is "
    f"{top_category}."
)

print(
    f"4. The product with the highest total sales is "
    f"{top_product}."
)

print(
    f"5. The city with the highest total sales is "
    f"{top_city}."
)

print(
    f"6. {top_payment} is the most commonly used payment method."
)

print(
    f"7. The average value of an order is "
    f"₹{average_sales:.2f}."
)

print(
    f"8. A total of {total_quantity} items were sold across "
    f"{number_of_orders} orders."
)
```
