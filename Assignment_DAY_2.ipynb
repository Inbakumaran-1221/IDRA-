```python

import random

# Set the range and number of attempts
lower_limit = 1
upper_limit = 100
max_attempts = 5

# Generate a random number
random_number = random.randint(lower_limit, upper_limit)

print("Welcome to the Number Guessing Game!")
print(f"I have selected a number between {lower_limit} and {upper_limit}.")
print(f"You have {max_attempts} attempts to guess it.\n")

# Give the user a fixed number of chances
for attempt in range(1, max_attempts + 1):
    guess = int(input(f"Attempt {attempt}: Enter your guess: "))

    if guess < random_number:
        print("Your guess is too low.\n")
    elif guess > random_number:
        print("Your guess is too high.\n")
    else:
        print("Correct! You guessed the number!")
        print(f"You got it in {attempt} attempt(s).")
        break
else:
    print("Sorry, you have used all your attempts.")
    print(f"The correct number was {random_number}.")

```
