```python
import pandas as pd

# Load the dataset
file_name = "Day7_Student_Dataset (1).csv"

df = pd.read_csv(file_name)

# Display the first 5 rows
print("First 5 rows:")
display(df.head())


# Display the last 5 rows
print("Last 5 rows:")
display(df.tail())


# Display random samples
print("Random 5 rows:")
display(df.sample(5))


# Check the shape of the dataset
print("Dataset Shape:")
print(df.shape)

print("\nNumber of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])


# Display column names
print("\nColumns in the dataset:")
print(df.columns)


# Display the index
print("\nDataset Index:")
print(df.index)


# Display data types
print("\nData Types:")
print(df.dtypes)


# Display basic information about the dataset
print("\nDataset Information:")
df.info()


# Generate descriptive statistics
print("\nDescriptive Statistics:")
display(df.describe())


# Check for missing values
print("\nMissing Values:")
print(df.isnull().sum())


# Check the total number of missing values
print("\nTotal Missing Values:")
print(df.isnull().sum().sum())


# Display basic information about each column
print("\nColumn Information:")
for column in df.columns:
    print(f"{column}: {df[column].dtype}")


# Basic observations
print("\n===== Dataset Summary =====")
print(f"Total number of rows: {df.shape[0]}")
print(f"Total number of columns: {df.shape[1]}")

print("\nFeatures available:")
for column in df.columns:
    print("-", column)

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
missing_values = df.isnull().sum()

for column, value in missing_values.items():
    if value > 0:
        print(f"{column}: {value} missing value(s)")

if missing_values.sum() == 0:
    print("No missing values found in the dataset.")


# Check unique courses
print("\nCourses available:")
print(df["Course"].unique())


# Check unique cities
print("\nCities available:")
print(df["City"].unique())
```
